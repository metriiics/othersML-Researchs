```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from matplotlib.colors import ListedColormap

from Perceptron import Perceptron
from ovr import OneVsRestClassifier
```

```python
df = pd.read_csv("iris.csv", encoding="utf-8", index_col="Unnamed: 0")
```

```python
X = df[["0", "2"]].values

class_ = {"Iris-setosa": 0, "Iris-versicolor": 1, "Iris-virginica": 2}

y = np.array([class_[i] for i in df["4"]])
y
```

```python
sns.scatterplot(df, x=df["0"], y=df["2"], hue=df["4"])
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
ppn = Perceptron(eta=0.01, n_iter=5000)

ovr = OneVsRestClassifier(ppn)
ovr.fit(X, y)
```

```python
df.loc[len(df)] = [5.1, 1.0, 'new_val']
df.loc[len(df)] = [5.5, 4.0, 'new_val']
df.loc[len(df)] = [7.5, 6.8, 'new_val']
```

```python
sns.scatterplot(df, x=df["0"], y=df["2"], hue=df["4"])
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
